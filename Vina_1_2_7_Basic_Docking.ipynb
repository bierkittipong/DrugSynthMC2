This notebook demonstrates a **basic molecular docking workflow** using the AutoDock Vina executable inside **Google Colab**.  
It provides a clean, reproducible pipeline suitable for **undergraduate workshops** and for users who want to run docking without installing anything locally.

In this example, a small-molecule ligand is prepared, converted to PDBQT, and docked into a rigid receptor.  
Multiple poses are generated, Vina scores are extracted and saved to CSV, and the best-scoring pose is visualized using py3Dmol or uploaded to **Proteins.plus** for interactive inspection.

The structure of this notebook follows the logic of the official basic docking example:
- Prepare ligand and receptor  
- Define a docking box  
- Run Vina  
- Extract individual poses  
- Analyze Vina scores  
- Visualize the complex  
- (Optional) Perform a short MD relaxation of the best pose  

A pose that closely resembles the experimentally observed binding mode is typically found among the top-ranked solutions.

# 📦 Box 4: Receptor Preparation

---

## 📖 What it does:
Prepares the **receptor** (protein target):

**Steps:**
1. 📥 **Downloads protein** from PDB database (e.g., PDB ID: 1M17)
2. 🔧 **Repairs structure** using PDBFixer:
   - Adds missing internal residues
   - Adds missing atoms
   - Removes non-standard residues
   - Removes water and other small molecules
3. ✂️ **Selects receptor atoms** (e.g., "chain A")
4. 🎯 **Defines docking box** (the search space where ligand will dock)
5. ⚡ **Adds hydrogens** using OpenBabel
6. 📄 **Creates PDBQT file** for Vina

---

## 💡 Why we need it:
Proteins from PDB databases often have:
- ❌ Missing atoms or residues
- ❌ Multiple chains (we usually want one)
- ❌ Water molecules and ions (we remove these)
- ❌ No hydrogens (we add them)

This cell cleans and prepares the protein for docking!

---

## 🎯 Analogy:
*Like preparing a lock before testing keys - you need to remove rust, make sure all parts are present, and clean out any debris.*

---

## 👀 What you'll see:
1. **Download message** for PDB file
2. **PDBFixer report:**
   - Number of missing residues added
   - Missing atoms added
   - Heterogens removed
3. **Selection confirmation** (e.g., "✓ Receptor atoms written")
4. **Docking box parameters:**
   - Center position (x, y, z coordinates)
   - Box size (width, height, depth in Ångströms)
5. **3D visualization** showing:
   - Receptor protein (cartoon representation)
   - Reference ligand (if present - magenta sticks)
   - Docking box (wireframe cube)

---

## ⚠️ Important Note:
**PDBFixer Limitations:**

✅ **Can fix:**
- Internal missing residues (between two existing residues)
- Missing heavy atoms
- Non-standard amino acids

❌ **Cannot fix:**
- Missing N-terminal residues (start of protein)
- Missing C-terminal residues (end of protein)
- Very long missing loops

If you see warnings about missing terminal regions, this is **normal** - it's a limitation of all automatic repair tools!

For complete reconstruction, use:
- **AlphaFold2/ColabFold**
- **MODELLER**
- **SWISS-MODEL**

---


````python
# @title # 1.1 Receptor Preparation
# @markdown Prepares receptor PDBQT and docking box from a PDB structure.
# @markdown Co-crystal ligand (if detected) is used to auto-center the grid.
%%time

import os
import json
import numpy as np
from prody import *
import py3Dmol
import ipywidgets as widgets
from IPython.display import display
from google.colab import files


# =========================================================
# (0) Utilities
# =========================================================
def _assert_file_ok(path: str, min_bytes: int = 100, msg: str = ""):
    """Raise if file is missing or suspiciously small."""
    if not os.path.exists(path) or os.path.getsize(path) < min_bytes:
        raise ValueError(msg or f"❌ File missing/too small: {path}")

def _safe_resname(x):
    return (x or "").strip()


# =========================================================
# (1) Choose / Load PDB
# =========================================================
source_mode = "Download from RCSB"  # @param ["Upload (use my file)", "Download from RCSB"]
pdb_token   = "3wy1"                # @param {type:"string"}

print("="*60)
print("(1) Loading PDB")
print("="*60)

if source_mode == "Upload (use my file)":
    print("📂 Upload your PDB file (protein complex).")
    uploaded = files.upload()
    if not uploaded:
        raise ValueError("❌ No file uploaded.")
    raw_pdb   = list(uploaded.keys())[0]
    pdb_token = os.path.splitext(os.path.basename(raw_pdb))[0]
    print(f"✓ Using uploaded file: {raw_pdb}")
else:
    raw_pdb = f"{pdb_token}.pdb"
    print(f"⬇ Downloading: {pdb_token}")
    !curl -s "https://files.rcsb.org/download/{pdb_token}.pdb" -o "{raw_pdb}"
    _assert_file_ok(raw_pdb, 200, f"❌ Download failed or file too small: {pdb_token}")
    print(f"✓ Downloaded: {raw_pdb}")

atoms_from_pdb = parsePDB(raw_pdb)
if atoms_from_pdb is None:
    raise ValueError("❌ ProDy parsePDB failed.")
print(f"✓ Total atoms in structure: {atoms_from_pdb.numAtoms()}")


# =========================================================
# (2) Co-crystal Ligand Detection → LIG.pdb + Grid Center
# =========================================================
print("\n" + "="*60)
print("(2) Co-crystal Ligand Detection")
print("="*60)

ligand_resname_hint   = "AUTO"  # @param {type:"string"}
exclude_ions_solvents = "HOH,WAT,DOD,SOL,NA,CL,K,CA,MG,ZN,MN,FE,CU,CO,NI,CD,HG"

# @markdown ---
# @markdown ### Ligand Filter Options
exclude_glycans   = True   # @param {type:"boolean"}
exclude_cofactors = True  # @param {type:"boolean"}

# Curated glycan / sugar residue names (common PDB HETATM entries)
_GLYCAN_NAMES = {
    "NAG", "BMA", "MAN", "FUC", "GAL", "GLC", "SIA", "NGA",
    "FUL", "GLA", "BGC", "A2G", "LAT", "MAL", "CEL", "SUC",
    "TRE", "GCS", "NDG", "NGC",
}

# Curated cofactors / buffers / crystallographic additives
_COFACTOR_NAMES = {
    "ATP", "ADP", "AMP", "GTP", "GDP", "GMP",
    "NAD", "NAP", "NDP", "FAD", "FMN",
    "HEM", "HEC", "HEA",
    "GOL", "PEG", "EDO", "MPD", "PGE", "PG4",
    "SO4", "PO4", "SUL", "PHO",
    "IHP", "TTP", "CTP", "UTP",
    "COA", "SAM", "SAH",
    "EPE", "MES", "TRS", "ACT", "ACY",
}

# Path to sidecar JSON that bridges the widget async gap
_LIG_CACHE = "/content/.dock_state_lig.json"


def build_candidate_list(atoms, exclude_csv, exclude_glycans=True, exclude_cofactors=False):
    """
    Return list of dicts for each unique hetero residue after all exclusion filters.
    Each dict: {resname, chain, resid, n_atoms, atoms}
    """
    het = atoms.select("hetero and not water")
    if het is None:
        return []

    excluded = {x.strip() for x in exclude_csv.split(",") if x.strip()}
    if exclude_glycans:
        excluded |= _GLYCAN_NAMES
    if exclude_cofactors:
        excluded |= _COFACTOR_NAMES

    candidates = []
    for res in het.getHierView().iterResidues():
        rn = _safe_resname(res.getResname())
        if rn in excluded:
            continue
        candidates.append({
            "resname": rn,
            "chain":   (res.getChid() or "").strip(),
            "resid":   res.getResnum(),
            "n_atoms": res.numAtoms(),
            "atoms":   res,
        })
    return candidates


def _label(c):
    """Dropdown display label: 'CFA:B:602 (13 atoms)'"""
    return f"{c['resname']}:{c['chain']}:{c['resid']} ({c['n_atoms']} atoms)"


def _sel_str(c):
    """ProDy selection string for a candidate dict."""
    return f"resname {c['resname']} and resid {c['resid']} and chain {c['chain']}"


def _finalise_ligand(chosen):
    """
    Populate shared ligand state and persist to sidecar JSON.
    Called for both auto-select (0/1 candidate) and widget confirm (>1).
    The JSON bridges the Colab widget async gap so Section (3) can recover state.
    """
    global ligand_atoms, ligand_selection, ligand_detected
    global ligand_pdb_path, proposed_cx, proposed_cy, proposed_cz

    sel = atoms_from_pdb.select(_sel_str(chosen))
    if sel is None or sel.numAtoms() == 0:
        raise ValueError(f"❌ Re-selection failed for: {_label(chosen)}")

    ligand_atoms     = sel
    ligand_selection = _sel_str(chosen)
    ligand_detected  = True
    ligand_pdb_path  = "LIG.pdb"

    writePDB(ligand_pdb_path, ligand_atoms)
    _assert_file_ok(ligand_pdb_path, 100, "❌ Failed to write LIG.pdb")

    proposed_cx, proposed_cy, proposed_cz = calcCenter(ligand_atoms)

    # Persist so Section (3) can load state after async widget confirm
    with open(_LIG_CACHE, "w") as _f:
        json.dump({
            "ligand_selection": ligand_selection,
            "ligand_pdb_path":  ligand_pdb_path,
            "ligand_detected":  True,
            "proposed_cx":      float(proposed_cx),
            "proposed_cy":      float(proposed_cy),
            "proposed_cz":      float(proposed_cz),
        }, _f)

    print(f"✅ Ligand finalised: {_label(chosen)}")
    print(f"   Selection : '{ligand_selection}'")
    print(f"   Atoms     : {ligand_atoms.numAtoms()}")
    print(f"✓  Wrote: {ligand_pdb_path}")
    print(f"📍 Proposed grid center: ({proposed_cx:.3f}, {proposed_cy:.3f}, {proposed_cz:.3f})")


# ── Initialise shared state ───────────────────────────────────────────────────
ligand_atoms     = None
ligand_selection = None
ligand_detected  = False
ligand_pdb_path  = None
proposed_cx = proposed_cy = proposed_cz = None

# ── Build candidates ──────────────────────────────────────────────────────────
all_candidates = build_candidate_list(
    atoms_from_pdb,
    exclude_csv       = exclude_ions_solvents,
    exclude_glycans   = exclude_glycans,
    exclude_cofactors = exclude_cofactors,
)

# ── Apply resname hint ────────────────────────────────────────────────────────
hint = (ligand_resname_hint or "").strip()
if hint and hint != "AUTO":
    hinted = [c for c in all_candidates if c["resname"] == hint]
    if hinted:
        print(f"ℹ️  Resname hint '{hint}' matched {len(hinted)} residue(s) — "
              f"{'auto-selecting' if len(hinted) == 1 else 'pre-selecting first in dropdown'}.")
        candidates = hinted
    else:
        print(f"⚠️  Resname hint '{hint}' matched nothing after filtering. "
              f"Falling back to all {len(all_candidates)} candidate(s).")
        candidates = all_candidates
else:
    candidates = all_candidates

print(f"   Candidates after filtering: {len(candidates)}")

# ── Collapse by resname: if only 1 unique residue name survives filtering,
#    auto-select the first chain (prefer A) regardless of how many chains exist.
unique_resnames = {c["resname"] for c in candidates}

if len(candidates) == 0:
    # ── 0 candidates ──────────────────────────────────────────────────────────
    print("⚠️  No ligand candidates detected after filtering.")
    print("   Remedies:")
    print("   • Set exclude_glycans=False or exclude_cofactors=False")
    print("   • Set ligand_resname_hint to your ligand's three-letter code")
    print("   • Use Grid Center mode 'No – enter XYZ manually' in Section (3)")

elif len(candidates) == 1 or len(unique_resnames) == 1:
    # ── 1 candidate, OR multiple chains of the same residue name ─────────────
    # Sort: chain A first, then ascending chain letter; pick index 0.
    candidates_sorted = sorted(candidates, key=lambda c: (c["chain"] != "A", c["chain"]))
    chosen = candidates_sorted[0]

    if len(candidates) > 1:
        skipped = [_label(c) for c in candidates_sorted[1:]]
        print(f"✅ Single residue name '{chosen['resname']}' across "
              f"{len(candidates)} chain(s) → auto-selecting first chain "
              f"({chosen['chain'] or '?'}).")
        print(f"   Skipped : {', '.join(skipped)}")
    else:
        print("✅ Single candidate → auto-selecting.")

    _finalise_ligand(chosen)

else:
    # ── >1 distinct residue names → interactive dropdown ─────────────────────
    print(f"ℹ️  {len(unique_resnames)} distinct ligand residue names found "
          f"({len(candidates)} total entries) — please choose one below.")

    candidates_sorted = sorted(candidates, key=lambda c: (c["chain"] != "A", -c["n_atoms"]))
    options = [_label(c) for c in candidates_sorted]

    dropdown = widgets.Dropdown(
        options     = options,
        value       = options[0],
        description = "Ligand:",
        layout      = widgets.Layout(width="400px"),
        style       = {"description_width": "60px"},
    )
    confirm_btn = widgets.Button(
        description  = "✔ Confirm selection",
        button_style = "success",
        layout       = widgets.Layout(width="200px"),
    )
    out = widgets.Output()

    def _on_confirm(_btn):
        out.clear_output()
        chosen = candidates_sorted[options.index(dropdown.value)]
        with out:
            _finalise_ligand(chosen)

    confirm_btn.on_click(_on_confirm)

    display(widgets.VBox([
        widgets.HTML(
            "<b>Multiple distinct HETATM ligand types detected.<br>"
            "Select the co-crystal ligand and click Confirm, "
            "then re-run the Section (3) cell:</b>"
        ),
        dropdown,
        confirm_btn,
        out,
    ]))
    print("\n⏳ Click '✔ Confirm selection' above, then run the Section (3) cell.")

# =========================================================
# (3) Grid Center
# =========================================================
print("\n" + "="*60)
print("(3) Grid Center")
print("="*60)

# ── Recover ligand state from sidecar JSON if widget hasn't fired yet ─────────
# Colab widget callbacks execute asynchronously (after the cell finishes),
# so ligand_detected may still be False when this code runs.
# _finalise_ligand() always writes _LIG_CACHE; we reload it here as a fallback.
if not ligand_detected and os.path.exists(_LIG_CACHE):
    try:
        with open(_LIG_CACHE) as _f:
            _ls = json.load(_f)
        if _ls.get("ligand_detected"):
            ligand_detected  = True
            ligand_selection = _ls["ligand_selection"]
            ligand_pdb_path  = _ls["ligand_pdb_path"]
            proposed_cx      = _ls["proposed_cx"]
            proposed_cy      = _ls["proposed_cy"]
            proposed_cz      = _ls["proposed_cz"]
            ligand_atoms     = parsePDB(ligand_pdb_path)  # reload for downstream use
            print(f"♻️  Ligand state recovered from cache: '{ligand_selection}'")
            print(f"   Grid center: ({proposed_cx:.3f}, {proposed_cy:.3f}, {proposed_cz:.3f})")
    except Exception as _e:
        print(f"⚠️  Could not load ligand cache ({_e}). Proceeding without it.")
# ─────────────────────────────────────────────────────────────────────────────

# @markdown ---
# @markdown ### Grid Center Mode
confirm_auto_grid = "Yes (use detected ligand)"  # @param ["Yes (use detected ligand)", "No – enter XYZ manually", "No – enter atom selection (resname / resid)"]

# @markdown ---
# @markdown ### Option A — Manual XYZ
# @markdown *(only used when mode = "No – enter XYZ manually")*
manual_center_x = 0.0  # @param {type:"number"}
manual_center_y = 0.0  # @param {type:"number"}
manual_center_z = 0.0  # @param {type:"number"}

# @markdown ---
# @markdown ### Option B — Atom Selection
# @markdown *(only used when mode = "No – enter atom selection")*
# @markdown Use ProDy syntax, e.g.:
# @markdown - `resname ATP` — by ligand residue name
# @markdown - `resname ATP and chain A` — ligand in a specific chain
# @markdown - `resid 84 86 134 and chain A` — specific protein residues
# @markdown - `resid 84 to 100 and chain B` — residue range
manual_atom_selection = "resname LIG and chain A"  # @param {type:"string"}

# ── Resolve center ────────────────────────────────────────────────────────────
if confirm_auto_grid == "Yes (use detected ligand)":
    if not ligand_detected:
        print("❌ STRICT MODE: Co-crystal ligand was NOT detected.")
        print("👉 ACTION REQUIRED:")
        print("   • If you just clicked Confirm in the widget above, re-run this cell.")
        print("   • Or switch mode to 'No – enter XYZ manually' and fill in X/Y/Z.")
        print("   • Or switch mode to 'No – enter atom selection' and enter a resname/resid.")
        raise ValueError("Auto-grid requested but no ligand was detected. See instructions above.")

    center_x = float(proposed_cx)
    center_y = float(proposed_cy)
    center_z = float(proposed_cz)
    print(f"✅ Using AUTO grid center from co-crystal ligand '{ligand_selection}'.")

elif confirm_auto_grid == "No – enter atom selection (resname / resid)":
    sel_str = manual_atom_selection.strip()
    if not sel_str:
        raise ValueError("❌ Atom selection is empty. Fill in 'manual_atom_selection'.")

    ref_atoms = atoms_from_pdb.select(sel_str)
    if ref_atoms is None or ref_atoms.numAtoms() == 0:
        raise ValueError(
            f"❌ Atom selection returned no atoms: '{sel_str}'\n"
            "   Check resname / resid / chain and try again."
        )

    center_x, center_y, center_z = calcCenter(ref_atoms)
    print(f"✅ Grid center from atom selection: '{sel_str}'")
    print(f"   Atoms matched : {ref_atoms.numAtoms()}")

    ligand_pdb_path = "LIG_ref.pdb"
    writePDB(ligand_pdb_path, ref_atoms)
    ligand_atoms    = ref_atoms
    ligand_detected = True
    print(f"✓  Reference atoms written: {ligand_pdb_path}")

else:  # Manual XYZ
    center_x = float(manual_center_x)
    center_y = float(manual_center_y)
    center_z = float(manual_center_z)

    if abs(center_x) < 1e-12 and abs(center_y) < 1e-12 and abs(center_z) < 1e-12:
        print("❌ Manual XYZ appears unset (all zeros).")
        print("👉 Fill in manual_center_x / manual_center_y / manual_center_z")
        raise ValueError("Manual center is 0,0,0. Please set valid coordinates.")

    print(f"🛠️  Using MANUAL grid center (XYZ).")

# ── Final validation ──────────────────────────────────────────────────────────
if np.any(np.isnan([center_x, center_y, center_z])):
    raise ValueError("❌ Grid center contains NaN values.")

print(f"📌 Final grid center: ({center_x:.3f}, {center_y:.3f}, {center_z:.3f})")


# =========================================================
# (4) Write example SMILES file (for downstream ligand prep)
# =========================================================
_ex_smi_content = """\
C1C(OC2=CC=CC=C2C1=O)C3=CC=CC=C3	Flavanone
C1=CC(=CC=C1C2=CC(=O)C3=C(C=C(C=C3O2)O)O)O	Apigenin
C1=CC=C(C=C1)C2=CC(=O)C3=C(O2)C=C(C(=C3O)O)O	Baicalein
C1=CC(=C(C=C1C2=CC(=O)C3=C(O2)C=C(C(=C3O)O)O)O)O	Luteolin
COC1=C(C=CC(=C1)C2=C(C(=O)C3=C(O2)C(=C(C=C3O)OC)OC)OC)O	Ternatin
"""
_ex_smi_path = "ex_smi.smi"
if os.path.exists(_ex_smi_path):
    print(f"⚠️  '{_ex_smi_path}' already exists — skipping write.")
else:
    with open(_ex_smi_path, "w") as f:
        f.write(_ex_smi_content)
    print(f"✓ '{_ex_smi_path}' written (example ligands for downstream use).")


# =========================================================
# (5) Receptor Atom Selection → receptor PDB
# =========================================================
print("\n" + "="*60)
print("(5) Receptor Selection")
print("="*60)

receptor_selection = (
    f"not ({ligand_selection}) and not water"
    if ligand_detected else "not water"
)

receptor_atoms = atoms_from_pdb.select(receptor_selection)
if receptor_atoms is None or receptor_atoms.numAtoms() == 0:
    raise ValueError(
        "❌ Receptor selection returned no atoms.\n"
        f"   Selection: {receptor_selection}"
    )

receptor_pdb = f"{pdb_token}_receptor_atoms.pdb"
writePDB(receptor_pdb, receptor_atoms)
_assert_file_ok(receptor_pdb, 200, "❌ Failed to write receptor PDB")
print(f"✓ Receptor PDB: {receptor_pdb}  ({receptor_atoms.numAtoms()} atoms)")


# =========================================================
# (6) Box Size (Å)
# =========================================================
size_x = 16  # @param {type:"slider", min:10, max:40, step:2}
size_y = 16  # @param {type:"slider", min:10, max:40, step:2}
size_z = 16  # @param {type:"slider", min:10, max:40, step:2}


# =========================================================
# (7) Receptor: Add H → PDBQT (OpenBabel)
# =========================================================
print("\n" + "="*60)
print("(7) Receptor PDBQT Preparation")
print("="*60)

prepare_output = "rec"  # @param {type:"string"}
receptor_fh    = f"{prepare_output}.pdb"
receptor_pdbqt = f"{prepare_output}.pdbqt"

# ── Metal ions that choke OpenBabel's Gasteiger / PDBQT writer ───────────────
_METAL_RESNAMES = {"MG", "ZN", "CA", "MN", "FE", "CU", "CO", "NI", "CD", "HG", "NA", "K"}
_METAL_CHARGES  = {"MG": 2.0, "ZN": 2.0, "CA": 2.0, "MN": 2.0, "FE": 3.0,
                   "CU": 2.0, "CO": 2.0, "NI": 2.0, "CD": 2.0, "HG": 2.0,
                   "NA": 1.0, "K":  1.0}

# ── (a) Strip metals from receptor PDB before OpenBabel ──────────────────────
metal_lines  = []
clean_lines  = []

with open(receptor_pdb) as _f:
    for line in _f:
        record = line[:6].strip()
        if record in ("ATOM", "HETATM"):
            resname = line[17:20].strip().upper()
            if resname in _METAL_RESNAMES:
                metal_lines.append(line)
                continue
        clean_lines.append(line)

receptor_pdb_nometal = receptor_pdb.replace(".pdb", "_nometal.pdb")
with open(receptor_pdb_nometal, "w") as _f:
    _f.writelines(clean_lines)

if metal_lines:
    print(f"⚠️  Stripped {len(metal_lines)} metal atom(s) before OpenBabel conversion:")
    for _l in metal_lines:
        print(f"   {_l.rstrip()}")
else:
    print("ℹ️  No metal atoms found — proceeding normally.")

# ── (b) Add H + convert to PDBQT (metal-free) ────────────────────────────────
print("⏳ Adding hydrogens...")
!obabel "{receptor_pdb_nometal}" -O "{receptor_fh}" -h 2>/dev/null
_assert_file_ok(receptor_fh, 200, "❌ Failed to generate H-added receptor PDB")
print(f"✓ H-added PDB: {receptor_fh}")

print("⏳ Converting to PDBQT...")
!obabel "{receptor_fh}" -O "{receptor_pdbqt}" -xr --partialcharge gasteiger 2>/dev/null
_assert_file_ok(receptor_pdbqt, 200, f"❌ Receptor PDBQT generation failed: {receptor_pdbqt}")
print(f"✓ Receptor PDBQT: {receptor_pdbqt}")

# ── (c) Re-inject metal ions into PDBQT with correct charges ─────────────────
if metal_lines:
    pdbqt_lines = open(receptor_pdbqt).readlines()
    pdbqt_lines = [l for l in pdbqt_lines if l.strip() != "END"]  # strip END

    injected = 0
    for line in metal_lines:
        resname = line[17:20].strip().upper()
        # Parse PDB columns
        try:
            serial = int(line[6:11])
            name   = line[12:16].strip()
            chain  = line[21] if len(line) > 21 else "A"
            resid  = int(line[22:26])
            x      = float(line[30:38])
            y      = float(line[38:46])
            z      = float(line[46:54])
        except ValueError:
            print(f"⚠️  Could not parse metal line — skipping: {line.rstrip()}")
            continue

        charge = _METAL_CHARGES.get(resname, 0.0)
        atom_type = resname.capitalize()  # e.g. "Mg", "Zn", "Ca"

        pdbqt_line = (
            f"HETATM{serial:5d} {name:<4s} {resname:<3s} {chain}{resid:4d}    "
            f"{x:8.3f}{y:8.3f}{z:8.3f}  1.00  0.00    {charge:+.3f} {atom_type}\n"
        )
        pdbqt_lines.append(pdbqt_line)
        injected += 1

    pdbqt_lines.append("END\n")
    with open(receptor_pdbqt, "w") as _f:
        _f.writelines(pdbqt_lines)

    print(f"✅ Re-injected {injected} metal atom(s) into {receptor_pdbqt}")
    import subprocess
    check = subprocess.run(["grep", "-E", "MG|ZN|CA|MN|FE|CU", receptor_pdbqt],
                           capture_output=True, text=True)
    if check.stdout:
        print("   Verified in PDBQT:")
        for _l in check.stdout.strip().split("\n"):
            print(f"   {_l}")


# =========================================================
# (8) Docking Box PDB + Vina Config File
# =========================================================
print("\n" + "="*60)
print("(8) Docking Box & Vina Config")
print("="*60)

box_pdb   = f"{prepare_output}.box.pdb"
configTXT = f"{prepare_output}.box.txt"

def write_box_pdb(filename, cx, cy, cz, sx, sy, sz):
    """Write 8-corner wireframe box as PDB HETATM records."""
    hx, hy, hz = sx / 2, sy / 2, sz / 2
    corners = [
        (cx-hx, cy-hy, cz-hz), (cx+hx, cy-hy, cz-hz),
        (cx-hx, cy+hy, cz-hz), (cx+hx, cy+hy, cz-hz),
        (cx-hx, cy-hy, cz+hz), (cx+hx, cy-hy, cz+hz),
        (cx-hx, cy+hy, cz+hz), (cx+hx, cy+hy, cz+hz),
    ]
    with open(filename, "w") as f:
        for i, (x, y, z) in enumerate(corners, 1):
            f.write(f"HETATM{i:5d}  C   BOX A   1    {x:8.3f}{y:8.3f}{z:8.3f}  1.00  0.00           C\n")
        f.write(
            "CONECT    1    2    3    5\nCONECT    2    1    4    6\n"
            "CONECT    3    1    4    7\nCONECT    4    2    3    8\n"
            "CONECT    5    1    6    7\nCONECT    6    2    5    8\n"
            "CONECT    7    3    5    8\nCONECT    8    4    6    7\n"
        )

write_box_pdb(box_pdb, center_x, center_y, center_z, size_x, size_y, size_z)
_assert_file_ok(box_pdb, 100, "❌ Failed to write box PDB")
print(f"✓ Box PDB: {box_pdb}")

with open(configTXT, "w") as f:
    f.write(
        f"center_x = {center_x:.4f}\n"
        f"center_y = {center_y:.4f}\n"
        f"center_z = {center_z:.4f}\n"
        f"size_x = {size_x}\n"
        f"size_y = {size_y}\n"
        f"size_z = {size_z}\n"
    )
_assert_file_ok(configTXT, 50, "❌ Failed to write Vina config")
print(f"✓ Vina config: {configTXT}")


# =========================================================
# (9) 3D Visualization
# =========================================================
print("\n" + "="*60)
print("(9) Visualization")
print("="*60)

box_style      = "stick"    # @param ["stick", "line", "surface"]
receptor_style = "cartoon"  # @param ["cartoon", "surface", "line", "stick"]
ligand_style   = "stick"    # @param ["stick", "line", "sphere"]

def visualize_docking_setup(receptor_pdb=None, box_pdb=None, ligand_pdb=None,
                             box_style="stick", receptor_style="cartoon", ligand_style="stick"):
    """Build a py3Dmol view with receptor, docking box, and optional co-crystal ligand."""
    _styles = {
        "stick":   {"stick": {}},
        "line":    {"line": {}},
        "sphere":  {"sphere": {"scale": 0.3}},
        "surface": {"surface": {"opacity": 0.4}},
        "cartoon": {"cartoon": {"color": "spectrum", "opacity": 0.6}},
    }
    def _style(kind): return _styles.get(kind, {"stick": {}})

    view = py3Dmol.view(width=600, height=400)   # fixed viewport size
    view.setBackgroundColor("white")
    idx  = 0

    for pdb_path, style_key, extra in [
        (box_pdb,      box_style,      {}),
        (receptor_pdb, receptor_style, {}),
        (ligand_pdb,   ligand_style,   {"colorscheme": "magentaCarbon"}),
    ]:
        if pdb_path and os.path.exists(pdb_path):
            with open(pdb_path) as f:
                view.addModel(f.read(), "pdb")
            s = _style(style_key)
            if extra and "stick" in s:
                s["stick"].update(extra)
            view.setStyle({"model": idx}, s)
            idx += 1

    view.zoomTo({"model": 0})
    view.zoom(0.65)
    return view

print("🎨 Rendering: receptor + docking box" +
      (" + co-crystal ligand" if ligand_detected else ""))

visualize_docking_setup(
    receptor_pdb=receptor_fh,
    box_pdb=box_pdb,
    ligand_pdb=ligand_pdb_path,
    box_style=box_style,
    receptor_style=receptor_style,
    ligand_style=ligand_style,
).show()


# =========================================================
# (10) Persist state → Section 2 can run without re-running 1.1
# =========================================================
_state_1 = {
    "receptor_pdbqt":  receptor_pdbqt,
    "receptor_fh":     receptor_fh,
    "configTXT":       configTXT,
    "box_pdb":         box_pdb,
    "pdb_token":       pdb_token,
    "center_x":        center_x,
    "center_y":        center_y,
    "center_z":        center_z,
    "size_x":          size_x,
    "size_y":          size_y,
    "size_z":          size_z,
    "ligand_pdb_path": ligand_pdb_path,
}
with open("/content/.dock_state_1.json", "w") as _f:
    json.dump(_state_1, _f)
print("💾 Session state saved → /content/.dock_state_1.json")

# =========================================================
# Summary
# =========================================================
print("\n" + "="*60)
print("📋 PREPARATION SUMMARY")
print("="*60)
print(f"  PDB token      : {pdb_token}")
print(f"  Receptor PDBQT : {receptor_pdbqt}")
print(f"  Vina config    : {configTXT}")
print(f"  Grid center    : ({center_x:.3f}, {center_y:.3f}, {center_z:.3f})")
print(f"  Box size       : {size_x} × {size_y} × {size_z} Å")
print(f"  Co-crystal LIG : {ligand_pdb_path or 'None (not detected)'}")
print("="*60)
````

# 📦 Box 5: Ligand Preparation

---

## 📖 What it does:
Prepares your **ligand** (the small molecule that will dock into the protein):

**Input methods:**
1. **SMILES string** - a text code representing the molecule (e.g., "CCO" for ethanol)
2. **File upload** - upload a .pdb, .mol2, or .sdf file

**Processing steps:**
1. 🤖 **Predicts pKa** using AI (how acidic/basic the molecule is)
2. ⚗️ **Adjusts protonation** at pH 7.4 (adds/removes H atoms appropriately)
3. 🔧 **Generates 3D structure** using RDKit
4. ⚡ **Optimizes geometry** (makes the structure realistic)
5. 📄 **Creates PDBQT file** for AutoDock Vina

## 💡 Why we need it:
Docking programs need molecules in a specific format (PDBQT) with correct protonation. This is like preparing ingredients before cooking - you need them at the right pH and in the right format!

---

## 🎯 Analogy:
*Like preparing a key before trying it in a lock - you need the right shape, the right size, and the right orientation.*

---

## 🌟 Real-world example:
**Drug discovery:** Before testing if a drug molecule binds to a protein, scientists must prepare it at the pH found in the body (around pH 7.4). Different pH = different charges = different binding!

---

## 👀 What you'll see:
1. **Predicted pKa value** (e.g., "🧮 Estimated pKa: 4.52")
2. **Formal charge** at pH 7.4 (e.g., "charge = 0" or "charge = -1")
3. **2D structure** - flat drawing of your molecule
4. **3D structure** - rotatable 3D visualization
5. **Files created:**
   - `ELR.pdbqt` - ready for docking
   - `ELR_scrubbed.sdf` - for visualization

---

## ⚠️ Important Note:
**SMILES codes** are like molecular zip codes:
- "C" = carbon atom
- "O" = oxygen atom  
- "CCO" = ethanol (alcohol)
- Numbers indicate ring structures
Example: "C1=CC=NC=C1C=NO" represents a complex drug molecule

````python
# @title # 1.2 Ligand Preparation (< 5s)
# @markdown In this step, the ligand is protonation-adjusted (pH-aware using Dimorphite-DL with a simple carboxylate fallback),
# @markdown a 3D conformer is generated, and a ligand PDBQT file suitable for AutoDock Vina is created.
# @markdown pKaNET (pKaPredict) is used to estimate pKa of the input ligand, and both **2D** and **3D** structures are shown.

%%time

import os
import sys
from pathlib import Path
import py3Dmol

# ============================================================
# 0. Fix Meeko compatibility with new RDKit (rdkit.six patch)
# ============================================================
try:
    from rdkit import six as rdkit_six
except ImportError:
    from io import StringIO
    from types import ModuleType
    import rdkit

    rdkit_six = ModuleType("six")
    rdkit_six.StringIO = StringIO
    rdkit_six.PY3 = True

    rdkit.six = rdkit_six
    sys.modules["rdkit.six"] = rdkit_six
    print("✓ Applied rdkit.six compatibility patch")

# ============================================================
# 1. Imports
# ============================================================
from rdkit import Chem
from rdkit.Chem import AllChem
from meeko import MoleculePreparation
from IPython.display import display

# RDKit Draw for 2D
try:
    from rdkit.Chem import Draw
except Exception as e:
    Draw = None
    print(f"⚠ RDKit Draw not available: {e}")

# pKaPredict (ML-based pKa, "pKaNET-style")
import pkapredict
from pkapredict import load_model, smiles_to_rdkit_descriptors, predict_pKa
print("pKaPredict version:", getattr(pkapredict, "__version__", "unknown"))

# Load pKa model once
try:
    pkanet_model
except NameError:
    pkanet_model = load_model()
    print("✓ pKaPredict model loaded (ML-based pKa engine)")

def predict_pka_pkanet(smiles_str: str) -> float:
    """Predict pKa using pKaNET/pKaPredict for a single SMILES (API-robust)."""
    smi = smiles_str.strip()
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        raise ValueError("❌ RDKit could not parse SMILES for pKa prediction.")
    # Handle old/new signatures of smiles_to_rdkit_descriptors
    try:
        desc = smiles_to_rdkit_descriptors([smi])
    except TypeError:
        desc = smiles_to_rdkit_descriptors([smi], descriptor_names=None)
    return float(predict_pKa(pkanet_model, desc)[0])

# Try Dimorphite-DL (for pH-aware protonation)
try:
    from dimorphite_dl import protonate_smiles
    HAVE_DIMORPHITE = True
    print("✓ dimorphite_dl detected: using pH-aware protonation")
except ImportError:
    HAVE_DIMORPHITE = False
    print("⚠ dimorphite_dl not found: using simple carboxylate rule only")

# ============================================================
# 2. User input
# ============================================================
# @markdown **Input SMILES (RDKit-canonical SMILES)**
ligand_Smiles = "COCCOC1=C(C=C2C(=C1)C(=NC=N2)NC3=CC=CC(=C3)C#C)OCCOC"  # @param {type:"string"}

# @markdown **Target pH**
pH = 7.4  # @param {type:"number"}

# @markdown **Output ligand PDBQT file name**
ligandPDBQT = "ELR.pdbqt"  # @param {type:"string"}

ligandName = ligandPDBQT.replace(".pdbqt", "")
ligandSDF  = f"{ligandName}_scrubbed.sdf"

print(f"Input SMILES: {ligand_Smiles}")
print(f"Target pH: {pH}")

# Optional: predict pKa with pKaNET for information
try:
    pka_est = predict_pka_pkanet(ligand_Smiles)
    print(f"🧮 Estimated pKa (pKaNET): {pka_est:.2f}")
except Exception as e:
    print(f"⚠ pKa prediction (pKaNET) failed: {e}")
    pka_est = None

# ============================================================
# 3. Protonation adjustment (Dimorphite-DL + simple fallback)
# ============================================================
def adjust_carboxylate_protonation(smiles: str, pH_value: float) -> str:
    """
    Simple pH-based rule (fallback if Dimorphite-DL unavailable or fails):
      - If pH >= 7 → deprotonate CO2H → CO2-
      - Otherwise keep input SMILES.
    Only handles standard carboxylic acids: C(=O)OH.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError("❌ RDKit failed to parse input SMILES in adjust_carboxylate_protonation.")

    # Below pH ~7, keep user-defined protonation
    if pH_value < 7.0:
        return smiles

    # SMARTS: C(=O)O-H  → matches (C, O_doublebond, O_hydroxyl)
    acid_pattern = Chem.MolFromSmarts("[CX3](=O)[OX2H1]")
    matches = mol.GetSubstructMatches(acid_pattern)

    if not matches:
        # No CO2H found
        return smiles

    emol = Chem.RWMol(mol)

    for (c_idx, o_double_idx, o_hydroxyl_idx) in matches:
        o_atom = emol.GetAtomWithIdx(o_hydroxyl_idx)

        # Only modify neutral –CO2H
        if o_atom.GetFormalCharge() == 0:
            # Deprotonate: CO2H → CO2–
            o_atom.SetFormalCharge(-1)
            o_atom.SetNumExplicitHs(0)
            o_atom.SetNoImplicit(True)

    newmol = emol.GetMol()
    Chem.SanitizeMol(newmol)
    new_smiles = Chem.MolToSmiles(newmol)
    return new_smiles

def adjust_protonation_with_dimorphite(smiles: str, pH_value: float) -> str:
    """
    Use Dimorphite-DL to generate a pH-adjusted protonation state.
    Falls back to simple carboxylate rule if Dimorphite gives no result.
    """
    if not HAVE_DIMORPHITE:
        return adjust_carboxylate_protonation(smiles, pH_value)

    try:
        variants = protonate_smiles(
            smiles,
            ph_min=pH_value,
            ph_max=pH_value,
            max_variants=1,
        )
        if not variants:
            print("⚠ Dimorphite-DL returned no variants, using simple carboxylate rule.")
            return adjust_carboxylate_protonation(smiles, pH_value)
        return variants[0]
    except Exception as e:
        print(f"⚠ Dimorphite-DL protonation failed ({e}), using simple carboxylate rule.")
        return adjust_carboxylate_protonation(smiles, pH_value)

prot_smiles = adjust_protonation_with_dimorphite(ligand_Smiles, pH)

# report protonation-adjusted SMILES and formal charge
mol_charge = Chem.MolFromSmiles(prot_smiles)
if mol_charge is None:
    raise ValueError("❌ RDKit failed to parse protonation-adjusted SMILES.")
total_charge = Chem.GetFormalCharge(mol_charge)

print(f"Protonation-adjusted SMILES (Dimorphite-DL / fallback): {prot_smiles}")
print(f"Estimated formal charge at pH {pH}: {total_charge}")

# ============================================================
# 4. Build 3D ligand with RDKit
# ============================================================
mol = Chem.MolFromSmiles(prot_smiles)
if mol is None:
    raise ValueError("❌ RDKit failed to parse protonation-adjusted SMILES.")

mol = Chem.AddHs(mol)

# Robust embedding (ETKDGv3/ETKDG + fallback to random coords)
try:
    try:
        params = AllChem.ETKDGv3()
    except AttributeError:
        params = AllChem.ETKDG()
    params.randomSeed = 42  # reproducible
    embed_result = AllChem.EmbedMolecule(mol, params)
except Exception as e:
    print(f"⚠ ETKDG embedding failed ({e}), trying simple call")
    embed_result = AllChem.EmbedMolecule(mol, randomSeed=42)

if embed_result == -1 or mol.GetNumConformers() == 0:
    print("⚠ Standard embedding failed, trying random coordinates")
    embed_result2 = AllChem.EmbedMolecule(mol, useRandomCoords=True, randomSeed=42)
    if embed_result2 == -1 or mol.GetNumConformers() == 0:
        raise ValueError("❌ 3D embedding failed for this ligand (no conformer)")

# Energy minimization (MMFF)
try:
    if AllChem.MMFFHasAllMoleculeParams(mol):
        converged = AllChem.MMFFOptimizeMolecule(mol, maxIters=500)
        if converged == 0:
            print("✓ MMFF optimization converged")
        else:
            print(f"⚠ MMFF optimization did not fully converge (code: {converged})")
    else:
        print("⚠ MMFF parameters unavailable, using UFF instead")
        converged = AllChem.UFFOptimizeMolecule(mol, maxIters=500)
except Exception as e:
    print(f"⚠ Geometry optimization failed ({e}), continuing with embedded structure")

# ============================================================
# 5. Write SDF (for visualization / debugging)
# ============================================================
with Chem.SDWriter(ligandSDF) as w:
    w.write(mol)
print(f"✓ SDF written: {os.path.abspath(ligandSDF)}")

# ============================================================
# 6. Prepare ligand PDBQT with Meeko
# ============================================================
preparator = MoleculePreparation()
preparator.prepare(mol)
pdbqt_string = preparator.write_pdbqt_string()

with open(ligandPDBQT, "w") as f:
    f.write(pdbqt_string)
print(f"✓ PDBQT written: {os.path.abspath(ligandPDBQT)}")

print(f"\n✓ Ligand prepared: {ligandName}")

# ============================================================
# 7. Visualization: 2D (RDKit) + 3D (py3Dmol)
# ============================================================
# ---- 2D depiction from protonation-adjusted SMILES ----
if Draw is not None:
    try:
        mol2d = Chem.MolFromSmiles(prot_smiles)
        if mol2d is not None:
            AllChem.Compute2DCoords(mol2d)
            img = Draw.MolToImage(mol2d, size=(300, 300))
            print("\n🧬 2D structure (protonation-adjusted SMILES):")
            display(img)
        else:
            print("⚠ Could not parse prot_smiles for 2D drawing.")
    except Exception as e:
        print(f"⚠ 2D drawing failed: {e}")
else:
    print("⚠ RDKit Draw not available, skipping 2D view.")

# ---- 3D depiction from SDF via py3Dmol ----
try:
    view = py3Dmol.view(width=600, height=400)
    with open(ligandSDF, "r") as f:
        sdf_content = f.read()
    view.addModel(sdf_content, "sdf")
    view.setBackgroundColor("white")
    view.setStyle({"stick": {"colorscheme": "yellowCarbon"}})
    view.zoomTo()
    view.show()
except Exception as e:
    print(f"⚠ Visualization error: {e}")
````

3. Searches for best binding positions within the box
4. Generates multiple poses (typically 9-20)
5. Scores each pose (more negative = better binding)

**Key parameter:**
- **Exhaustiveness** (4-64): How thorough the search is
  - 8 = standard (fast, usually sufficient)
  - 32 = thorough (slower but more comprehensive)
  - 64 = very thorough (much slower)

This is the **main event!** Vina tries millions of different positions and orientations to find how your ligand molecule best fits into the protein binding site.

````python
# @title # 4. Interactive Pose Selection, Visualization & Extraction
# @markdown Interactively browse docked poses, visualize them with the receptor and box,
# @markdown and save the **currently selected pose** to an SDF file for further analysis.
# @markdown
# @markdown **Tips**
# @markdown - Use the slider to change the pose index.
# @markdown - The selected pose is saved to `output_sdf` on each update.
# @markdown - You can upload `rec.pdb` and the selected SDF to Proteins.plus.

from rdkit import Chem
import py3Dmol
import ipywidgets as widgets
from IPython.display import display
import os

# ----------------------------------------------------------
# User input
# ----------------------------------------------------------
input_sdf  = "output.sdf"        # @param {type:"string"}
output_sdf = "selected_pose.sdf" # @param {type:"string"}

# Ensure extensions
if not input_sdf.endswith(".sdf"):
    input_sdf += ".sdf"
if not output_sdf.endswith(".sdf"):
    output_sdf += ".sdf"

# These come from previous cells:
#   receptor_fh     → prepared receptor PDB (rec.pdb)
#   prody_ligandPDB → reference ligand PDB (optional)
#   box_pdb         → docking box PDB
receptorPDB = receptor_fh
ligandPDB = ligand_pdb_path or ""
boxPDB      = box_pdb

# ----------------------------------------------------------
# Load poses from SDF
# ----------------------------------------------------------
if not os.path.exists(input_sdf):
    raise FileNotFoundError(f"❌ Docked SDF not found: {input_sdf}")

supplier = Chem.SDMolSupplier(input_sdf, sanitize=False)
mols = [m for m in supplier if m is not None]

if len(mols) == 0:
    raise ValueError(f"❌ No valid molecules found in {input_sdf}")

num_poses = len(mols)
print(f"🧮 Detected {num_poses} docked poses in {input_sdf}")

# ----------------------------------------------------------
# Build 3D viewer (receptor + box + all poses as frames)
# ----------------------------------------------------------
sdf_text = open(input_sdf, "rt").read()

v = py3Dmol.view(width=800, height=600)
v.setBackgroundColor("white")

model_index = 0

# Receptor
if receptorPDB and os.path.exists(receptorPDB):
    with open(receptorPDB) as f:
        v.addModel(f.read(), "pdb")
    v.setStyle({"model": model_index}, {
        "cartoon": {"color": "spectrum", "opacity": 0.6},
        "stick":   {"radius": 0.15, "opacity": 0.3},
    })
    model_index += 1

# Box
if boxPDB and os.path.exists(boxPDB):
    with open(boxPDB) as f:
        v.addModel(f.read(), "pdb")
    v.setStyle({"model": model_index}, {"line": {}})
    model_index += 1

# Reference ligand (optional)
if ligandPDB and os.path.exists(ligandPDB):
    with open(ligandPDB) as f:
        v.addModel(f.read(), "pdb")
    v.setStyle({"model": model_index}, {
        "stick": {"colorscheme": "magentaCarbon", "radius": 0.2},
    })
    model_index += 1

# Docked poses as frames (single model with many frames)
v.addModelsAsFrames(sdf_text)
poses_model_index = model_index  # remember index for poses

v.setStyle({"model": poses_model_index}, {
    "stick": {"colorscheme": "greenCarbon", "radius": 0.25},
})

v.zoomTo({'model':1})
v.rotate(90)
v.show()

print("\n📌 Model indices in viewer:")
print("  - receptor       → model 0 (if present)")
print("  - box            → model 1 (if present)")
print("  - reference lig. → model 2 (if present)")
print(f"  - docked poses   → model {poses_model_index} (animated frames)")

# ----------------------------------------------------------
# Interactive pose selector
# ----------------------------------------------------------
print("\n🔎 Use the slider to select a pose and write it to SDF:")

pose_slider = widgets.IntSlider(
    value=0,
    min=0,
    max=num_poses - 1,
    step=1,
    description="Pose #",
    continuous_update=False,
)

status_label = widgets.Label()

def update_pose(idx):
    idx = int(idx)

    # Update 3D viewer frame
    v.setFrame(idx)
    v.update()

    # Write selected pose to SDF
    sel_mol = mols[idx]
    w = Chem.SDWriter(output_sdf)
    w.write(sel_mol)
    w.close()

    status_label.value = (
        f"Currently showing pose {idx+1}/{num_poses} → "
        f"saved as '{output_sdf}'."
    )

widgets.interact(update_pose, idx=pose_slider)
display(status_label)

print("\n✅ Interactive pose selection ready.")
print("   Download 'rec.pdb' and the selected SDF file for external visualization.")
````

````python
# @title # 5. **Extract Vina Scores → Save CSV → Split Poses → Plot (All-in-One)**

import os
import pandas as pd
import matplotlib.pyplot as plt
from rdkit import Chem

# ======================================================
# Pipeline variables (already defined earlier)
# ======================================================
pdbqt_file = outputPDBQT                        # Vina output file (.pdbqt)
dock_sdf    = f"{dock_base}.sdf"                # Already exported using mk_export
out_csv     = f"{dock_base}_vina_scores.csv"    # Output table
pose_dir    = "DOCKPOSES"

print(f"📄 Reading docking results from: {pdbqt_file}")
print(f"📄 Reading SDF poses from:        {dock_sdf}")

os.makedirs(pose_dir, exist_ok=True)

# ======================================================
# STEP 1 — Split SDF poses into individual files (PDB + SDF)
# ======================================================
confs = Chem.SDMolSupplier(dock_sdf, sanitize=False)
pose_files = []

for i, mol in enumerate(confs, start=1):
    if mol is None:
        continue

    pdb_file = os.path.join(pose_dir, f"{dock_base}_{i}.pdb")
    sdf_file = os.path.join(pose_dir, f"{dock_base}_{i}.sdf")

    Chem.MolToPDBFile(mol, pdb_file)
    w = Chem.SDWriter(sdf_file); w.write(mol); w.close()

    pose_files.append((pdb_file, sdf_file))

print(f"✓ Extracted {len(pose_files)} poses → saved in '{pose_dir}/'")

# ======================================================
# STEP 2 — Extract Vina scores from PDBQT
# ======================================================
data = []
current_model = None

with open(pdbqt_file, "r") as f:
    for line in f:
        line = line.strip()

        if line.startswith("MODEL"):
            try:
                current_model = int(line.split()[1])
            except:
                current_model = None

        elif line.startswith("REMARK VINA RESULT:"):
            parts = line.split()
            try:
                score   = float(parts[3])
                rmsd_lb = float(parts[4])
                rmsd_ub = float(parts[5])
            except:
                continue

            data.append({
                "model": current_model,
                "vina_score_kcal_mol": score,
                "rmsd_lb": rmsd_lb,
                "rmsd_ub": rmsd_ub
            })

# ======================================================
# STEP 3 — Save CSV table
# ======================================================
df = pd.DataFrame(data)
df = df.sort_values("vina_score_kcal_mol").reset_index(drop=True)
df.to_csv(out_csv, index=False)

print(f"✓ Docking score table saved → {os.path.abspath(out_csv)}")
print(df)

# ======================================================
# STEP 4 — Plot score ranking
# ======================================================
plt.figure(figsize=(6, 4))
plt.bar(df["model"], df["vina_score_kcal_mol"],
        color="skyblue", edgecolor="black")
plt.gca().invert_yaxis()
plt.xlabel("MODEL index")
plt.ylabel("Vina Score (kcal/mol)")
plt.title("Vina Docking Scores per Pose")
plt.tight_layout()
plt.show()
````

````python
# @title # 6. **Package Results → Download**
# @markdown ### ⚙️ Options
# @markdown **`zip_files`** — Pack everything into a single `.zip` before downloading.
# @markdown > 💡 **iPad / mobile users**: set `zip_files = False`.
# @markdown > Colab will download each file individually, which is more reliable
# @markdown > on devices that cannot easily open ZIP archives.
# @markdown
# @markdown Files are **grouped by docking engine** (VINA / GNINA / UNIDOCK / VINAXB)
# @markdown so outputs never get mixed together.

import shutil
import os
import re
import zipfile
from google.colab import files

# ============================================================================
# CONFIGURATION
# ============================================================================

zip_files  = True                   # @param {type:"boolean"}
zip_name   = "vina_results.zip"     # @param {type:"string"}

# ============================================================================
# RESOLVE PIPELINE VARIABLES
# ============================================================================

_engine    = str(globals().get("dock_engine", "VINA")).upper().strip()
_dock_base = globals().get("dock_base", "dock")
_pdb_token = globals().get("pdb_token", "PDB")

_ENGINE_LABELS = {
    "VINA":    "AutoDock_Vina",
    "VINAXB":  "Vina-XB",
    "UNIDOCK": "Uni-Dock",
    "GNINA":   "GNINA",
}
_ENGINE_EXTS = {
    "VINA":    (".pdbqt", ".sdf", ".log"),
    "VINAXB":  (".pdbqt", ".sdf", ".log"),
    "UNIDOCK": (".sdf",   ".log"),
    "GNINA":   (".sdf",   ".log"),
}
_eng_label = _ENGINE_LABELS.get(_engine, _engine)
_eng_exts  = _ENGINE_EXTS.get(_engine, (".sdf",))

# ========= PIPELINE-AWARE SETTINGS ========= #
folder_to_zip = "DOCKPOSES"
rec_pdb       = globals().get("receptor_fh",     None)   # e.g. "5NFP_receptorFH.pdb"
lig_pdb       = globals().get("prody_ligandPDB", None)   # e.g. "LIG.pdb"
vina_csv      = f"{_dock_base}_vina_scores.csv"          # from score-extraction cell
raw_pdb       = f"{_pdb_token}.pdb"                      # original PDB from RCSB
# ============================================ #

print("=" * 60)
print(f"  Engine  : {_engine}  ({_eng_label})")
print(f"  ZIP     : {'Yes — ' + zip_name if zip_files else 'No — individual files'}")
print("=" * 60)

# ============================================================================
# HELPERS
# ============================================================================

_stage = "ZIP_TMP"

def _reset_stage(folder):
    if os.path.exists(folder):
        shutil.rmtree(folder)
    os.makedirs(folder, exist_ok=True)

def safe_copy(src, dst_folder, label=None):
    tag = label or (os.path.basename(str(src)) if src else "<None>")
    if not src:
        print(f"  ✗ {tag} — not set")
        return
    if os.path.exists(str(src)):
        os.makedirs(dst_folder, exist_ok=True)
        shutil.copy(str(src), dst_folder)
        print(f"  ✓ {tag}")
    else:
        print(f"  ✗ {tag} — missing: {src} (skipped)")

def _subdir(*parts):
    p = os.path.join(_stage, *parts)
    os.makedirs(p, exist_ok=True)
    return p

# ============================================================================
# STAGE FILES — grouped by engine
# ============================================================================

_reset_stage(_stage)

# ── shared/ : structural files common to all engines ─────────────────────────
print(f"\n  📂 shared/")
_shared = _subdir("shared")
safe_copy(rec_pdb, _shared, "Receptor PDB")
safe_copy(lig_pdb, _shared, "Ligand PDB")
safe_copy(raw_pdb, _shared, f"Raw PDB ({_pdb_token})")

# ── <Engine>/ : all engine-specific outputs ───────────────────────────────────
print(f"\n  📂 {_eng_label}/")
_eng_dir = _subdir(_eng_label)

# Scores CSV
safe_copy(vina_csv, _subdir(_eng_label, "scores"), "Scores CSV")

# DOCKPOSES folder
_poses_dst = os.path.join(_eng_dir, folder_to_zip)
if os.path.exists(folder_to_zip):
    shutil.copytree(folder_to_zip, _poses_dst, dirs_exist_ok=True)
    n = sum(len(fs) for _, _, fs in os.walk(folder_to_zip))
    print(f"  ✓ {folder_to_zip}/ ({n} file(s))")
else:
    print(f"  ✗ {folder_to_zip}/ — folder not found (skipped)")

# ── README ────────────────────────────────────────────────────────────────────
with open(os.path.join(_stage, "README.txt"), "w") as f:
    f.write(f"Docking Results\n")
    f.write(f"Engine  : {_engine}  ({_eng_label})\n")
    f.write("=" * 40 + "\n\n")
    f.write("shared/          Receptor, ligand, raw PDB\n")
    f.write(f"{_eng_label}/\n")
    f.write(f"  scores/        {_dock_base}_vina_scores.csv\n")
    f.write(f"  {folder_to_zip}/     All docked poses\n")
print(f"  ✓ README.txt")

# ============================================================================
# COUNT STAGED FILES
# ============================================================================
_all_staged = [
    os.path.join(r, fn)
    for r, _, fns in os.walk(_stage)
    for fn in fns
]
print(f"\n  Total staged: {len(_all_staged)} file(s)")

# ============================================================================
# ZIP MODE
# ============================================================================
if zip_files:
    if os.path.exists(zip_name):
        os.remove(zip_name)

    with zipfile.ZipFile(zip_name, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for fp in _all_staged:
            zf.write(fp, os.path.relpath(fp, _stage))

    _zsize = os.path.getsize(zip_name)
    shutil.rmtree(_stage, ignore_errors=True)
    print(f"\n🎉 ZIP created: {zip_name}  ({_zsize/1024/1024:.1f} MB)")
    files.download(zip_name)

# ============================================================================
# NO-ZIP MODE — individual file downloads (recommended for iPad / mobile)
# ============================================================================
else:
    print("\n  ℹ️  Downloading files individually (iPad / mobile mode)...\n")
    for fp in sorted(_all_staged):
        rel = os.path.relpath(fp, _stage)
        print(f"  ⬇️  {rel}")
        files.download(fp)
    shutil.rmtree(_stage, ignore_errors=True)
    print(f"\n✅ Done — {len(_all_staged)} file(s) downloaded individually.")
````

````python
# @title # 1. **Cyclodextrin Hosts (PDB + optional PDBQT from GitHub)**
# @markdown Downloads PDB files for structure / visualization
# @markdown Downloads PDBQT only when available (for Vina docking)

import os, pathlib, urllib.request

BCD_DIR = pathlib.Path("BCD")
BCD_DIR.mkdir(exist_ok=True)

# Base GitHub URLs
BASE_PDB   = "https://raw.githubusercontent.com/nyelidl/host-guest/refs/heads/main/BCD"
BASE_PDBQT = "https://raw.githubusercontent.com/nyelidl/host-guest/refs/heads/main/BCD"

# ------------------------------------------------------------
# Host definitions
#   key  : display name
#   pdb  : required
#   pdbqt: optional (None if not available)
# ------------------------------------------------------------
hosts = {
    "BCD-qm": {
        "pdb":   "BCD-qm.pdb",
        "pdbqt": None,
    },
    "BCD": {
        "pdb":   "gBCD.pdb",
        "pdbqt": "gBCD.pdbqt",
    },
    "DMBCD": {
        "pdb":   "gDMBCD.pdb",
        "pdbqt": "gDMBCD.pdbqt",
    },
    "HPBCD": {
        "pdb":   "gHPBCD.pdb",
        "pdbqt": "gHPBCD.pdbqt",
    },
    "MBCD": {
        "pdb":   "gMBCD.pdb",
        "pdbqt": "gMBCD.pdbqt",
    },
    "6tetraHPBCD": {
        "pdb":   "g6tetraHPBCD.pdb",
        "pdbqt": "g6tetraHPBCD.pdbqt",
    },
    "ACD": {
        "pdb":   "gACD.pdb",
        "pdbqt": "gACD.pdbqt",
    },
    "YCD": {
        "pdb":   "gYCD.pdb",
        "pdbqt": "gYCD.pdbqt",
    },
}

# ------------------------------------------------------------
# Download helper
# ------------------------------------------------------------
def download(url, out_path):
    if out_path.exists():
        print(f"   ✓ {out_path.name} already exists")
        return
    try:
        urllib.request.urlretrieve(url, out_path)
        print(f"   ✓ Downloaded {out_path.name}")
    except Exception as e:
        print(f"   ⚠️ Failed to download {url}")
        print(f"     Reason: {e}")

# ------------------------------------------------------------
# Download loop
# ------------------------------------------------------------
for name, cfg in hosts.items():
    print(f"\n🔹 {name}")

    # ---- PDB (always required) ----
    pdb_file = cfg["pdb"]
    pdb_url  = f"{BASE_PDB}/{pdb_file}"
    pdb_out  = BCD_DIR / pdb_file
    print(f"   PDB:   {pdb_url}")
    download(pdb_url, pdb_out)

    # ---- PDBQT (optional) ----
    pdbqt_file = cfg["pdbqt"]
    if pdbqt_file:
        pdbqt_url = f"{BASE_PDBQT}/{pdbqt_file}"
        pdbqt_out = BCD_DIR / pdbqt_file
        print(f"   PDBQT:{pdbqt_url}")
        download(pdbqt_url, pdbqt_out)
    else:
        print("   PDBQT: — (not provided for this host)")

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------
print("\n✅ Cyclodextrin host download complete")
print("📂 Contents of ./BCD:")
for f in sorted(os.listdir(BCD_DIR)):
    print("  ", f)
````

````python
# @title # 1.1 Ligand Preparation (< 5s)
# @markdown In this step, the ligand is protonation-adjusted (pH-aware using Dimorphite-DL with a simple carboxylate fallback),
# @markdown a 3D conformer is generated, and a ligand PDBQT file suitable for AutoDock Vina is created.
# @markdown pKaNET (pKaPredict) is used to estimate pKa of the input ligand, and both **2D** and **3D** structures are shown.

%%time

import os
import sys
from pathlib import Path
import py3Dmol

# ============================================================
# 0. Fix Meeko compatibility with new RDKit (rdkit.six patch)
# ============================================================
try:
    from rdkit import six as rdkit_six
except ImportError:
    from io import StringIO
    from types import ModuleType
    import rdkit

    rdkit_six = ModuleType("six")
    rdkit_six.StringIO = StringIO
    rdkit_six.PY3 = True

    rdkit.six = rdkit_six
    sys.modules["rdkit.six"] = rdkit_six
    print("✓ Applied rdkit.six compatibility patch")

# ============================================================
# 1. Imports
# ============================================================
from rdkit import Chem
from rdkit.Chem import AllChem
from meeko import MoleculePreparation
from IPython.display import display

# RDKit Draw for 2D
try:
    from rdkit.Chem import Draw
except Exception as e:
    Draw = None
    print(f"⚠ RDKit Draw not available: {e}")

# pKaPredict (ML-based pKa, "pKaNET-style")
import pkapredict
from pkapredict import load_model, smiles_to_rdkit_descriptors, predict_pKa
print("pKaPredict version:", getattr(pkapredict, "__version__", "unknown"))

# Load pKa model once
try:
    pkanet_model
except NameError:
    pkanet_model = load_model()
    print("✓ pKaPredict model loaded (ML-based pKa engine)")

def predict_pka_pkanet(smiles_str: str) -> float:
    """Predict pKa using pKaNET/pKaPredict for a single SMILES (API-robust)."""
    smi = smiles_str.strip()
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        raise ValueError("❌ RDKit could not parse SMILES for pKa prediction.")
    # Handle old/new signatures of smiles_to_rdkit_descriptors
    try:
        desc = smiles_to_rdkit_descriptors([smi])
    except TypeError:
        desc = smiles_to_rdkit_descriptors([smi], descriptor_names=None)
    return float(predict_pKa(pkanet_model, desc)[0])

# Try Dimorphite-DL (for pH-aware protonation)
try:
    from dimorphite_dl import protonate_smiles
    HAVE_DIMORPHITE = True
    print("✓ dimorphite_dl detected: using pH-aware protonation")
except ImportError:
    HAVE_DIMORPHITE = False
    print("⚠ dimorphite_dl not found: using simple carboxylate rule only")

# ============================================================
# 2. User input
# ============================================================
# @markdown **Input SMILES (parent structure, approximate protonation)**
ligand_Smiles = "COCCOc1cc2c(cc1OCCOC)ncnc2Nc3cccc(c3)C#C"  # @param {type:"string"}

# @markdown **Target pH**
pH = 7.4  # @param {type:"number"}

# @markdown **Output ligand PDBQT file name**
ligandPDBQT = "ELR.pdbqt"  # @param {type:"string"}

ligandName = ligandPDBQT.replace(".pdbqt", "")
ligandSDF  = f"{ligandName}_scrubbed.sdf"

print(f"Input SMILES: {ligand_Smiles}")
print(f"Target pH: {pH}")

# Optional: predict pKa with pKaNET for information
try:
    pka_est = predict_pka_pkanet(ligand_Smiles)
    print(f"🧮 Estimated pKa (pKaNET): {pka_est:.2f}")
except Exception as e:
    print(f"⚠ pKa prediction (pKaNET) failed: {e}")
    pka_est = None

# ============================================================
# 3. Protonation adjustment (Dimorphite-DL + simple fallback)
# ============================================================
def adjust_carboxylate_protonation(smiles: str, pH_value: float) -> str:
    """
    Simple pH-based rule (fallback if Dimorphite-DL unavailable or fails):
      - If pH >= 7 → deprotonate CO2H → CO2-
      - Otherwise keep input SMILES.
    Only handles standard carboxylic acids: C(=O)OH.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError("❌ RDKit failed to parse input SMILES in adjust_carboxylate_protonation.")

    # Below pH ~7, keep user-defined protonation
    if pH_value < 7.0:
        return smiles

    # SMARTS: C(=O)O-H  → matches (C, O_doublebond, O_hydroxyl)
    acid_pattern = Chem.MolFromSmarts("[CX3](=O)[OX2H1]")
    matches = mol.GetSubstructMatches(acid_pattern)

    if not matches:
        # No CO2H found
        return smiles

    emol = Chem.RWMol(mol)

    for (c_idx, o_double_idx, o_hydroxyl_idx) in matches:
        o_atom = emol.GetAtomWithIdx(o_hydroxyl_idx)

        # Only modify neutral –CO2H
        if o_atom.GetFormalCharge() == 0:
            # Deprotonate: CO2H → CO2–
            o_atom.SetFormalCharge(-1)
            o_atom.SetNumExplicitHs(0)
            o_atom.SetNoImplicit(True)

    newmol = emol.GetMol()
    Chem.SanitizeMol(newmol)
    new_smiles = Chem.MolToSmiles(newmol)
    return new_smiles

def adjust_protonation_with_dimorphite(smiles: str, pH_value: float) -> str:
    """
    Use Dimorphite-DL to generate a pH-adjusted protonation state.
    Falls back to simple carboxylate rule if Dimorphite gives no result.
    """
    if not HAVE_DIMORPHITE:
        return adjust_carboxylate_protonation(smiles, pH_value)

    try:
        variants = protonate_smiles(
            smiles,
            ph_min=pH_value,
            ph_max=pH_value,
            max_variants=1,
        )
        if not variants:
            print("⚠ Dimorphite-DL returned no variants, using simple carboxylate rule.")
            return adjust_carboxylate_protonation(smiles, pH_value)
        return variants[0]
    except Exception as e:
        print(f"⚠ Dimorphite-DL protonation failed ({e}), using simple carboxylate rule.")
        return adjust_carboxylate_protonation(smiles, pH_value)

prot_smiles = adjust_protonation_with_dimorphite(ligand_Smiles, pH)

# report protonation-adjusted SMILES and formal charge
mol_charge = Chem.MolFromSmiles(prot_smiles)
if mol_charge is None:
    raise ValueError("❌ RDKit failed to parse protonation-adjusted SMILES.")
total_charge = Chem.GetFormalCharge(mol_charge)

print(f"Protonation-adjusted SMILES (Dimorphite-DL / fallback): {prot_smiles}")
print(f"Estimated formal charge at pH {pH}: {total_charge}")

# ============================================================
# 4. Build 3D ligand with RDKit
# ============================================================
mol = Chem.MolFromSmiles(prot_smiles)
if mol is None:
    raise ValueError("❌ RDKit failed to parse protonation-adjusted SMILES.")

mol = Chem.AddHs(mol)

# Robust embedding (ETKDGv3/ETKDG + fallback to random coords)
try:
    try:
        params = AllChem.ETKDGv3()
    except AttributeError:
        params = AllChem.ETKDG()
    params.randomSeed = 42  # reproducible
    embed_result = AllChem.EmbedMolecule(mol, params)
except Exception as e:
    print(f"⚠ ETKDG embedding failed ({e}), trying simple call")
    embed_result = AllChem.EmbedMolecule(mol, randomSeed=42)

if embed_result == -1 or mol.GetNumConformers() == 0:
    print("⚠ Standard embedding failed, trying random coordinates")
    embed_result2 = AllChem.EmbedMolecule(mol, useRandomCoords=True, randomSeed=42)
    if embed_result2 == -1 or mol.GetNumConformers() == 0:
        raise ValueError("❌ 3D embedding failed for this ligand (no conformer)")

# Energy minimization (MMFF)
try:
    if AllChem.MMFFHasAllMoleculeParams(mol):
        converged = AllChem.MMFFOptimizeMolecule(mol, maxIters=500)
        if converged == 0:
            print("✓ MMFF optimization converged")
        else:
            print(f"⚠ MMFF optimization did not fully converge (code: {converged})")
    else:
        print("⚠ MMFF parameters unavailable, using UFF instead")
        converged = AllChem.UFFOptimizeMolecule(mol, maxIters=500)
except Exception as e:
    print(f"⚠ Geometry optimization failed ({e}), continuing with embedded structure")

# ============================================================
# 5. Write SDF (for visualization / debugging)
# ============================================================
with Chem.SDWriter(ligandSDF) as w:
    w.write(mol)
print(f"✓ SDF written: {os.path.abspath(ligandSDF)}")

# ============================================================
# 6. Prepare ligand PDBQT with Meeko
# ============================================================
preparator = MoleculePreparation()
preparator.prepare(mol)
pdbqt_string = preparator.write_pdbqt_string()

with open(ligandPDBQT, "w") as f:
    f.write(pdbqt_string)
print(f"✓ PDBQT written: {os.path.abspath(ligandPDBQT)}")

print(f"\n✓ Ligand prepared: {ligandName}")

# ============================================================
# 7. Visualization: 2D (RDKit) + 3D (py3Dmol)
# ============================================================
# ---- 2D depiction from protonation-adjusted SMILES ----
if Draw is not None:
    try:
        mol2d = Chem.MolFromSmiles(prot_smiles)
        if mol2d is not None:
            AllChem.Compute2DCoords(mol2d)
            img = Draw.MolToImage(mol2d, size=(300, 300))
            print("\n🧬 2D structure (protonation-adjusted SMILES):")
            display(img)
        else:
            print("⚠ Could not parse prot_smiles for 2D drawing.")
    except Exception as e:
        print(f"⚠ 2D drawing failed: {e}")
else:
    print("⚠ RDKit Draw not available, skipping 2D view.")

# ---- 3D depiction from SDF via py3Dmol ----
try:
    view = py3Dmol.view(width=600, height=400)
    with open(ligandSDF, "r") as f:
        sdf_content = f.read()
    view.addModel(sdf_content, "sdf")
    view.setBackgroundColor("white")
    view.setStyle({"stick": {"colorscheme": "yellowCarbon"}})
    view.zoomTo()
    view.show()
except Exception as e:
    print(f"⚠ Visualization error: {e}")
````

````python
# @title # 1.2 Host visualization + SIDE VIEW + Docking Grid Box + (optional) PDBQT generation
from pathlib import Path
import os

# ---------------------------
# User Settings (ONE host_name for both visualization & docking prep)
# ---------------------------
host_name = "gACD"  # @param ["BCD-qm","gBCD","gDMBCD","gHPBCD","gMBCD","g6tetraHPBCD","gACD","gYCD"]

center_x = 0.0      # @param {type:"number"}
center_y = 0.0      # @param {type:"number"}
center_z = 0.0      # @param {type:"number"}
size_x   = 20.0     # @param {type:"number"}
size_y   = 20.0     # @param {type:"number"}
size_z   = 30.0     # @param {type:"number"}

print(f"Selected host: {host_name}")

# ---------------------------
# File preparation
# ---------------------------
BCD_DIR = Path("BCD")
pdb_in = BCD_DIR / f"{host_name}.pdb"
sdf_out = Path(f"{host_name}.sdf")

# Final receptor filename expected by later docking step
receptor_pdbqt = Path("host.pdbqt")

# If a pre-made PDBQT exists in ./BCD, prefer it (more reliable than generating)
pdbqt_in = BCD_DIR / f"{host_name}.pdbqt"

if not pdb_in.exists():
    raise FileNotFoundError(f"{pdb_in} not found! Run the download cell first.")

# ---------------------------
# Ensure Open Babel exists
# ---------------------------
def have_obabel():
    return os.system("obabel -V > /dev/null 2>&1") == 0

if not have_obabel():
    print("Installing Open Babel (conda)...")
    # assumes conda is available from your previous install cell
    os.system("conda install -y -q -c conda-forge openbabel")

    if not have_obabel():
        raise RuntimeError("❌ Open Babel (obabel) not available after install.")

# ---------------------------
# Convert PDB → SDF (for visualization)
# ---------------------------
print("Converting PDB → SDF for visualization...")
os.system(f"obabel '{pdb_in}' -O '{sdf_out}' --addh > /dev/null 2>&1")

if not sdf_out.exists():
    raise RuntimeError("❌ Failed to create SDF. Check Open Babel installation or input PDB.")

# ---------------------------
# py3Dmol install
# ---------------------------
try:
    import py3Dmol
except ImportError:
    !pip install -q py3Dmol
    import py3Dmol

# ---------------------------
# Load SDF for visualization
# ---------------------------
with open(sdf_out) as f:
    sdf_block = f.read()

# ---------------------------
# Create viewer
# ---------------------------
view = py3Dmol.view(width=650, height=480)
view.addModel(sdf_block, "sdf")
view.setStyle({"stick": {"radius": 0.25}})

# ---------------------------
# Add GRID BOX (wireframe cube)
# ---------------------------
cx, cy, cz = center_x, center_y, center_z
sx, sy, sz = size_x/2, size_y/2, size_z/2

cube_vertices = [
    [cx-sx, cy-sy, cz-sz], [cx+sx, cy-sy, cz-sz],
    [cx+sx, cy+sy, cz-sz], [cx-sx, cy+sy, cz-sz],
    [cx-sx, cy-sy, cz+sz], [cx+sx, cy-sy, cz+sz],
    [cx+sx, cy+sy, cz+sz], [cx-sx, cy+sy, cz+sz]
]

edges = [
    (0,1),(1,2),(2,3),(3,0),
    (4,5),(5,6),(6,7),(7,4),
    (0,4),(1,5),(2,6),(3,7)
]

for i, j in edges:
    a = cube_vertices[i]
    b = cube_vertices[j]
    view.addLine({
        "start": {"x":a[0], "y":a[1], "z":a[2]},
        "end":   {"x":b[0], "y":b[1], "z":b[2]},
        "color": "green",
        "linewidth": 3
    })

# ---------------------------
# Set SIDE VIEW
# ---------------------------
view.rotate(90, "x")
view.zoomTo()
view.show()

print("✓ Side view + grid box visualized")

# ---------------------------
# PDBQT for docking (optional, correctness-first)
# ---------------------------
# Docking-ready (pre-validated) CD hosts in your repo:
# - gBCD, gDMBCD, gHPBCD usually have matching PDBQT files
# Anything else: do NOT silently generate unless you explicitly want a fallback method.
DOCKING_READY = {"gBCD", "gDMBCD", "gHPBCD"}

print("\n--- Docking receptor (PDBQT) preparation ---")

# Case 1) If PDBQT exists in ./BCD, use it (best)
if pdbqt_in.exists():
    receptor_pdbqt.write_text(pdbqt_in.read_text())
    print(f"✓ Using provided PDBQT: {pdbqt_in} → {receptor_pdbqt}")

# Case 2) Generate only for known docking-ready hosts (fallback)
elif host_name in DOCKING_READY:
    print(f"⚠️ No pre-made {pdbqt_in.name} found. Generating PDBQT with Open Babel (Gasteiger)...")
    cmd = f"obabel '{pdb_in}' -O '{receptor_pdbqt}' --partialcharge gasteiger --addh"
    rc = os.system(cmd + " > /dev/null 2>&1")
    if rc == 0 and receptor_pdbqt.exists():
        print(f"✓ Generated receptor PDBQT: {receptor_pdbqt}")
    else:
        raise RuntimeError("❌ Failed to create host.pdbqt. Check Open Babel output.")

# Case 3) Non-docking-ready hosts: skip with clear instruction
else:
    print(f"ℹ️ {host_name} does not have a provided PDBQT in ./BCD, and auto-generating PDBQT is skipped.")
    print("   Reason: carbohydrate atom typing/charges for docking are not reliably handled by a generic Gasteiger workflow.")
    print("   Action: use a pre-generated PDBQT (recommended) or explicitly enable a fallback method if you really want it.")
````

````python
# @title # 3. **Animate Docked Poses (Side View) — py3Dmol**

import py3Dmol
from pathlib import Path

# ===============================================
# Determine filenames
# ===============================================
host_pdb = f"BCD/{host_name}.pdb"
dock_sdf = f"{host_name}_{ligandName}_vina.sdf"

animation_interval_ms = 700  # @param {type:"slider", min:100, max:2000, step:100}

# ===============================================
# File checks
# ===============================================
print("📁 Checking files...")

if not Path(host_pdb).exists():
    raise FileNotFoundError(f"❌ Host PDB not found: {host_pdb}")
else:
    print(f"✓ Host PDB: {host_pdb}")

if not Path(dock_sdf).exists():
    raise FileNotFoundError(f"❌ Docked SDF not found: {dock_sdf}")
else:
    print(f"✓ Docked SDF: {dock_sdf}")

print("\n🎬 Building side-view animation...")

# ===============================================
# Viewer
# ===============================================
v = py3Dmol.view(width=850, height=600)
v.setBackgroundColor('white')

# --- Host CD structure ---
with open(host_pdb) as f:
    v.addModel(f.read(), 'pdb')

# Optional: show stick for host (can remove if you want only surface)
v.setStyle({'model': 0}, {
    'stick': {'colorscheme': 'cyanCarbon', 'radius': 0.16}
})

# --- Add mesh surface ---
v.addSurface(py3Dmol.MS, {
    "opacity": 0.35,
    "color": "lightgray",
    "surface": "SES",      # SES = solvent excluded surface (better for CD)
    "map": "mesh"          # <— mesh mode
}, {"model": 0})

v.zoomTo()

# --- Docked poses as animation frames ---
sdf_text = Path(dock_sdf).read_text()
v.addModelsAsFrames(sdf_text)

# Style for heavy atoms
v.setStyle({'model': 1, 'not': {'atom': 'H'}}, {
    'stick': {'colorscheme': 'greenCarbon', 'radius': 0.24}
})

# Hide all hydrogens
v.setStyle({'model': 1, 'atom': 'H'}, {})

# ===============================================
# Set SIDE VIEW
# ===============================================
# A cyclodextrin's axis is aligned with the Z axis in your PDB files.
# Rotate 90° around the X axis → looks from the side into the cavity.
v.rotate(90, 'x')

# ===============================================
# Animate poses
# ===============================================
v.animate({
    'interval': animation_interval_ms,
    'loop': 'forward'
})

v.zoomTo()
v.show()

print("\n✅ Side-view animation ready! Ligand poses shown in sequence.")
````


# 📦 Box 15: Extract Vina Scores → Save CSV → Split Poses → Plot

---

## 📖 What it does:
**Complete analysis package** for cyclodextrin docking:

1. **Reads Vina output** (binding energies)
2. **Extracts scores** for all poses
3. **Saves CSV table** (host_ligand_vina_scores.csv)
4. **Splits poses** into individual files
5. **Creates bar chart** showing ranking

**Files generated:**
- CSV table with all scores
- Folder `DOCKPOSES_gBCD_ELR_vina/` containing:
  - Individual PDB files (pose 1, 2, 3...)
  - Individual SDF files (pose 1, 2, 3...)

---

## 💡 Why we need it:
To **quantify and compare** different binding modes:
- Which insertion mode is strongest?
- How much variation between poses?
- Are there multiple good modes?
- Save data for publication/reports

---

## 👀 What you'll see:

**1) Console output:**
```
📄 Reading docking results from: gBCD_ELR_vina.pdbqt
📄 Reading SDF poses from: gBCD_ELR_vina.sdf
✓ Extracted 10 poses → saved in 'DOCKPOSES_gBCD_ELR_vina/'
```

````python
# @title # 4. **Extract Vina Scores → Save CSV → Split Poses → Plot (All-in-One)**

import os
import pandas as pd
import matplotlib.pyplot as plt
from rdkit import Chem
from pathlib import Path

# ======================================================
# Pipeline variables (from previous docking step)
# ======================================================
try:
    _ = host_name, ligandName
except NameError:
    raise NameError("host_name or ligandName not defined. Run docking cell first.")

dock_base   = f"{host_name}_{ligandName}_vina"
pdbqt_file  = f"{dock_base}.pdbqt"        # Vina output file (.pdbqt)
dock_sdf    = f"{dock_base}.sdf"          # SDF from docking cell (Open Babel)
out_csv     = f"{dock_base}_vina_scores.csv"
pose_dir    = f"DOCKPOSES_{dock_base}"

print(f"📄 Reading docking results from: {pdbqt_file}")
print(f"📄 Reading SDF poses from:      {dock_sdf}")

if not Path(pdbqt_file).exists():
    raise FileNotFoundError(f"❌ PDBQT file not found: {pdbqt_file}")

if not Path(dock_sdf).exists():
    raise FileNotFoundError(f"❌ SDF file not found: {dock_sdf}")

os.makedirs(pose_dir, exist_ok=True)

# ======================================================
# STEP 1 — Split SDF poses into individual files (PDB + SDF)
# ======================================================
confs = Chem.SDMolSupplier(dock_sdf, sanitize=False)
pose_files = []

for i, mol in enumerate(confs, start=1):
    if mol is None:
        continue

    pdb_file = os.path.join(pose_dir, f"{dock_base}_{i}.pdb")
    sdf_file = os.path.join(pose_dir, f"{dock_base}_{i}.sdf")

    Chem.MolToPDBFile(mol, pdb_file)
    w = Chem.SDWriter(sdf_file)
    w.write(mol)
    w.close()

    pose_files.append((pdb_file, sdf_file))

print(f"✓ Extracted {len(pose_files)} poses → saved in '{pose_dir}/'")

# ======================================================
# STEP 2 — Extract Vina scores from PDBQT
# ======================================================
data = []
current_model = None

with open(pdbqt_file, "r") as f:
    for line in f:
        line = line.strip()

        if line.startswith("MODEL"):
            try:
                current_model = int(line.split()[1])
            except Exception:
                current_model = None

        elif line.startswith("REMARK VINA RESULT:"):
            parts = line.split()
            try:
                score   = float(parts[3])
                rmsd_lb = float(parts[4])
                rmsd_ub = float(parts[5])
            except Exception:
                continue

            data.append({
                "model": current_model,
                "vina_score_kcal_mol": score,
                "rmsd_lb": rmsd_lb,
                "rmsd_ub": rmsd_ub
            })

if not data:
    raise RuntimeError("❌ No Vina RESULT lines found in PDBQT file.")

# ======================================================
# STEP 3 — Save CSV table
# ======================================================
df = pd.DataFrame(data)
df = df.sort_values("vina_score_kcal_mol").reset_index(drop=True)
df.to_csv(out_csv, index=False)

print(f"✓ Docking score table saved → {os.path.abspath(out_csv)}")
display(df)

# ======================================================
# STEP 4 — Plot score ranking
# ======================================================
plt.figure(figsize=(6, 4))
plt.bar(df["model"], df["vina_score_kcal_mol"])
plt.gca().invert_yaxis()
plt.xlabel("MODEL index")
plt.ylabel("Vina Score (kcal/mol)")
plt.title("Vina Docking Scores per Pose")
plt.tight_layout()
plt.show()
````
